### Задача 1.   

Пусть $X_1,...,X_n$ выборка из некоторого распределения $\mathsf{P}$, причем ${\sf D} X_1=\sigma^2<+\infty$, и $\sigma$ неизвестно.
Рассмотрим оценку $S^2 = \overline{X^2}-\overline{X}^2$ дисперсии $\sigma^2$.

**1.** Докажите, что $S^2=\frac1n\sum\limits_{i=1}^n\left(X_i-\overline{X}\right)^2$.

При n = 1 формула очевидна. Далее во всем задании рассматривается случай n > 1.

Рассмотрим произвольный индекс i. Найдем, с каким коэффицентом $X_i^2$ входит в $\overline{X^2}-\overline{X}^2$

В $\overline{X^2}$, $X_i^2$ входит в 1/n. В $\overline{X}^2 X_i^2$ входит в степени $\frac{1}{n^2}$. Значит суммарно $X_i^2$ входит с коэффицентом $\frac{1}{n} - \frac{1}{n^2} = \frac{n - 1}{n^2}$.

Найдем, с каким коэффицентом $X_i^2$ входит в $\frac1n\sum\limits_{j=1}^n\left(X_j-\overline{X}\right)^2$.

Коэффицент в сумме при j = i: $(X_i - \overline{X})^2$, $X_i^2$ входит с коэффицентом $(1 - 1/n)^2 = \frac{1 - 2n + n^2}{n^2}$.

При каждом j != i: в слагаемое $(X_j - \overline{X})^2$, $X_i$ входит с коэффицентом $\frac{1}{n^2}$. Таких слагаемых n - 1. Значит во все j != i $X_i^2$ входит с коэффицентом $(n - 1) / n^2$.

Суммируя, получаем: $\frac{n - 1}{n^2} + \frac{1 - 2n + n^2}{n^2} = \frac{n^2 - n}{n^2}$

Вспомним, что перед $\sum\limits_{j=1}^n\left(X_i-\overline{X}\right)^2$ есть коэффицент $\frac1n$, значит суммарно $X_i^2$ входит с коэффицентом $\frac{n^2-n}{n^2}\frac1n = \frac{n - 1}{n^2}$, то есть с таким же коэффицентом, что и в $\overline{X^2}-\overline{X}^2$.

Итого, мы выяснили, что $\forall i \in {1, 2, ..., n}$, $X_i^2$ входит в $\overline{X^2}-\overline{X}^2$ и в $\frac1n\sum\limits_{j=1}^n\left(X_j-\overline{X}\right)^2$ с одним и тем же коэффицентом.

Осталось рассмотреть, c каким коэффицентом входит $X_i X_j$ при i != j.

В $\overline{X^2}-\overline{X}^2$:
- В $\overline{X^2}$: с коэффицентом 0
- В $\overline{X}^2$: с коэффицентом $2 / n^2$

Откуда получаем, что в $\overline{X^2}-\overline{X}^2$, $X_i X_j$ входит с коэффицентом $-\frac{1}{n^2}$.

В $\frac1n\sum\limits_{k=1}^n\left(X_k-\overline{X}\right)^2$:
- При k != i и k != j в $(X_k - \overline{X})^2$: Входит с коэффицентом $2 / n^2$. Таких слагаемых n - 2 -> суммарный вклад: $\frac{n - 2}{n^2}$
- при k = i в $(X_i - \overline{X})^2$, $X_i X_j$ входит с: $-\frac{n - 1}{n^2} $
- при k = j аналогично случаю k = i.

Суммируя, получаем $-\frac{n - 1}{n^2} - \frac{n - 1}{n^2} + \frac{n - 2}{n^2} = -\frac1n$. Вспомним про коэффицент $\frac1n$ перед $\sum\limits_{k=1}^n\left(X_k-\overline{X}\right)^2$. Значит суммарный коэффицент будет $-\frac{1}{n^2}$

Так как это верно для $\forall i, j$ таких что i != j, то значит коэффиценты при всех таких $X_i X_j$ совпадают в обоих равенствах тоже.

Из того, что коэффиценты при $X_i^2$ и при $X_i X_j$ совпадают в обеих формулах, а одночленов другого вида в этих формулах нет, следует то, что формулы равны.

**2.** Посчитайте математическое ожидание случайной величины $S^2$. Является ли она несмещенной оценкой $\sigma^2$?

В пункте выше мы выяснили, что $\forall i \in {1, ..., n}$, $X_i^2$ входит с коэффицентом $\frac{n-1}{n^2}$. Всего n таких слагаемых, так как 1 <= i <= n.
Тогда их вклад в матожидание: $\mathbb{E}(\sum_{i=1}^n X_i^2) = \text{В силу линейности} = \sum_{i = 1}^n\mathbb{E}(\frac{n - 1}{n^2}X_i^2) = \text{В силу того, что } X_i^2 \text{ из одного распределения} = n \mathbb{E}(\frac{n - 1}{n^2}X_1^2) = \frac{n - 1}{n}\mathbb{E}(X_1^2)$

Для $\forall i, j \in {1, ..., n}$ при i != j, коэффицент при $X_i X_j$, равный $-\frac{1}{n^2}$ встретился n (n - 1) раз по всем i, j.

Тогда $\mathbb{E}(\sum_{i != j}(\frac{1}{n^2}X_i X_j)) = \text{По линейности математического ожидания} = \sum_{i != j}(\mathbb{E}(\frac{1}{n^2} X_i X_j)) = \text{В силу того, что } X_i \text{ из одного распределения}  = \frac{n (n - 1)}{n^2} \mathbb{E}(X_1 X_2) = \text{В силу независимости} X_i \text{ и } X_j = \frac{n - 1}{n}(\mathbb{E}(X_1))^2$

Значит $\mathbb{E}(S^2) = \frac{n - 1}{n}\left(\mathbb{E}(X_1^2) - (\mathbb{E}(X_1)))^2\right) = \frac{n - 1}{n}\mathbb{Var}(X_1)$

Вывод:
- Оценка является смещенной
- Что бы получить несмещенную оценку нужно рассмотреть величину $\frac{n}{n - 1}S^2$

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

sigma_true = 3.0
sigma2_true = sigma_true**2

sample_count = 500
sample_sizes = [10, 100, 500]

def estimator_1(samples):
    return np.var(samples, axis=1, ddof=0)

def estimator_2(samples):
    return np.var(samples, axis=1, ddof=1)

estimators = [
    (r"$\hat{\theta}_1=\frac{1}{n}\sum_{i=1}^n (X_i-\bar X)^2$", estimator_1, "tab:blue"),
    (r"$\hat{\theta}_2=\frac{1}{n-1}\sum_{i=1}^n (X_i-\bar X)^2$", estimator_2, "tab:orange"),
]

for sample_size in sample_sizes:
    samples = np.random.normal(
        loc=0,
        scale=sigma_true,
        size=(sample_count, sample_size)
    )

    results = []
    for label, estimator, color in estimators:
        theta_hats = estimator(samples)
        theta_mean = np.mean(theta_hats)
        theta_std = np.std(theta_hats, ddof=1)
        theta_bias = theta_mean - sigma2_true

        results.append({
            "label": label,
            "values": theta_hats,
            "mean": theta_mean,
            "std": theta_std,
            "bias": theta_bias,
            "color": color
        })

    fig, axes = plt.subplots(1, 3, figsize=(20, 5))
    ax_scatter, ax_hist, ax_box = axes

    # Scatter
    # Я делаю нормальный шум по оси OY.
    # Это нужно что бы точки не сливались
    for i, res in enumerate(results):
        y = i + np.random.normal(0, 0.04, size=sample_count)

        ax_scatter.scatter(
            res["values"],
            y,
            alpha=0.1,
            s=100,
            color=res["color"],
            label=res["label"]
        )

        ax_scatter.scatter(
            res["mean"],
            i,
            marker="*",
            s=200,
            color="white",
            edgecolors="black",
            zorder=5
        )

        ax_scatter.axvline(
            res["mean"],
            color=res["color"],
            linestyle="--",
            alpha=0.8
        )

    ax_scatter.axvline(
        sigma2_true,
        ls=":",
        color="grey",
        linewidth=2,
        label=r"Истинное $\sigma^2$"
    )
    ax_scatter.set_title(f"Отдельные оценки, n = {sample_size}")
    ax_scatter.set_yticks([0, 1])
    ax_scatter.set_yticklabels([r"$\hat{\theta}_1$", r"$\hat{\theta}_2$"])
    ax_scatter.set_xlabel("Значение оценки")
    ax_scatter.legend()

    # Гистограммы, дело в том, что scatter очень не информативен.
    # По нему сложно отличить синие от ораньжевых точек даже при n = 10
    bins = 30
    for res in results:
        ax_hist.hist(
            res["values"],
            bins=bins,
            density=True,
            alpha=0.5,
            color=res["color"],
            label=res["label"]
        )
        ax_hist.axvline(
            res["mean"],
            color=res["color"],
            linestyle="--",
            alpha=0.9
        )

    ax_hist.axvline(
        sigma2_true,
        ls=":",
        color="black",
        linewidth=2,
        label=r"Истинное $\sigma^2$"
    )
    ax_hist.set_title(f"Гистограммы, n = {sample_size}")
    ax_hist.set_xlabel("Значение оценки")
    ax_hist.set_ylabel("Плотность")
    ax_hist.legend()

    # Нарисовав гистограммы, я понял, что они почти идентичные, решил обавить boxplot
    box = ax_box.boxplot(
        [results[0]["values"], results[1]["values"]],
        vert=False,
        patch_artist=True,
        tick_labels=[r"$\hat{\theta}_1$", r"$\hat{\theta}_2$"]
    )

    for patch, res in zip(box["boxes"], results):
        patch.set_facecolor(res["color"])
        patch.set_alpha(0.35)

    ax_box.axvline(sigma2_true, ls=":", color="grey", linewidth=2)
    ax_box.set_title(f"Boxplot, n = {sample_size}")
    ax_box.set_xlabel("Значение оценки")

    stats_text = ""
    for idx, res in enumerate(results, start=1):
        stats_text += (
            f"Оценка {idx}:\n"
            f"mean = {res['mean']:.4f}\n"
            f"bias = {res['bias']:.4f}\n"
            f"std  = {res['std']:.4f}\n\n"
        )

    fig.text(
        0.86, 0.5,
        stats_text,
        fontsize=11,
        va="center",
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.9)
    )

    plt.suptitle(
        f"Сравнение оценок дисперсии при размере выборки n = {sample_size}",
        fontsize=14
    )
    plt.tight_layout(rect=[0, 0, 0.84, 0.95])
    plt.show()
```

**2.** Изучим поведение среднего оценок из первого пункта при росте размера выборки. Постройте график зависимости $\widetilde{\theta}$ от $n$ для двух типов оценок. Для вычисления зависимости нужно один раз сгенерировать выборки из пункта 1 достаточно большого размера (не более 500) и посчитать оценки по подвыборкам, используя <a href="https://thetahat.ru/courses/python/03"  target="_top">функции</a> из `numpy`. Использовать циклы, а так же функции, разворачивающиеся в цикл (например, `np.vectorize`), запрещено.

**Решение:**

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(52)

sigma_true = 3.0
sigma2_true = sigma_true**2

sample_count = 500
max_n = 130

samples = np.random.normal(
    loc=0,
    scale=sigma_true,
    size=(sample_count, max_n)
)

n_values = np.arange(1, max_n + 1)

samples_squared = samples**2

sum_x_prefix = np.cumsum(samples, axis=1)
sum_x_squared_over_n = (sum_x_prefix**2) / n_values
sum_x2_prefix = np.cumsum(samples_squared, axis=1)

# 1-я оценка
theta1_values = (sum_x2_prefix - sum_x_squared_over_n) / np.arange(1, max_n + 1)
theta1_mean = np.mean(theta1_values, axis=0)

# 2-я оценка
theta2_values = (sum_x2_prefix - sum_x_squared_over_n)[:, 1:] / np.arange(1, max_n)
theta2_mean = np.mean(theta2_values, axis=0)

n_values_2 = np.arange(2, max_n + 1)

plt.figure(figsize=(10, 5))

plt.plot(
    n_values,
    theta1_mean,
    label=r"$\hat{\theta}_1=\frac{1}{n}\sum_{i=1}^n (X_i-\bar X)^2$"
)

plt.plot(
    n_values_2,
    theta2_mean,
    label=r"$\hat{\theta}_2=\frac{1}{n-1}\sum_{i=1}^n (X_i-\bar X)^2$"
)

plt.axhline(
    sigma2_true,
    ls=":",
    color="grey",
    linewidth=2,
    label=r"Истинное $\sigma^2$"
)

plt.xlabel("Размер выборки")
plt.ylabel("Значение оценки")
plt.title(r"Зависимость средних оценок $\hat{\theta}_1$ и $\hat{\theta}_2$ от размера выборки")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
def plot_histories(histories: List[Dict[str, Dict[str, list]]], names: List[str]) -> None:
    """
    Визуализирует лосс и точность для нескольких моделей на одном графике.

    Параметры:
        histories (List[Dict[str, Dict[str, list]]]): Список историй обучения моделей.
            Каждая история должна содержать ключи "loss" и "acc" с подключами "train" и "val".
        names (List[str]): Список названий моделей для легенды.

    Возвращает:
        None
    """
    # Устанавливаем стиль "darkgrid" только для текущего графика
    with sns.axes_style("darkgrid"):
        # Цвета для графиков
        colors = ["darkblue", "lightcoral", "limegreen", "sandybrown",  "purple"]

        # Создаем фигуру с двумя подграфиками
        fig, axs = plt.subplots(1, 2, figsize=(14, 5))

        # Определяем количество эпох для отображения (минимальное среди всех историй)
        epochs = np.min([len(h["loss"]["train"]) for h in histories])

        # Построение графиков для каждой модели
        for i, (history, name) in enumerate(zip(histories, names)):
            # График лосса
            axs[0].set_title("Лосс")
            axs[0].plot(
                history["loss"]["train"][:epochs],
                label=f"{name}",
                lw=2,
                c=colors[i],
                ls="--",
            )
            axs[0].plot(
                history["loss"]["val"][:epochs],
                lw=1.5,
                c=colors[i],
            )
            axs[0].set_xlabel("Номер эпохи")

            # График точности
            axs[1].set_title("Точность классификации")
            axs[1].plot(
                history["acc"]["train"][:epochs], label=f"{name}", lw=2, c=colors[i], ls="--"
            )
            axs[1].plot(history["acc"]["val"][:epochs], lw=1.5, c=colors[i])
            axs[1].set_xlabel("Номер эпохи")
            axs[1].legend()

        # Создаем фиктивные линии для легенды (train и val)
        dummy_lines = [
            axs[0].plot([], [], c="black", lw=2, ls="--")[0],
            axs[0].plot([], [], c="black", lw=1.5)[0],
        ]

        # Добавляем легенду для типов линий (train и val)
        for i in range(2):
            legend = axs[i].legend(loc=3 - i)  # Легенда для моделей
            axs[i].legend(dummy_lines, ["train", "val"], loc=4)  # Легенда для типов линий
            axs[i].add_artist(legend)  # Добавляем обе легенды на график

        plt.show()


def print_epoch(epoch, num_epochs, history, t):
    """
    Функция для вывода информации про эпоху.
    :param epoch: номер эпохи
    :param num_epochs: общее количество эпох
    :param history: (dict) accuracy и loss на обучении и валидации ("история" обучения)
    :param t: время эпохи в секундах
    """
    clear_output(wait=True)
    print("Эпоха {} из {}, затраченное время {:.3f} сек.".format(epoch + 1, num_epochs, t))
    print("  лосс на обучении: \t{:.6f}".format(history["loss"]["train"][-1]))
    print("  лосс на валидации: \t{:.6f}".format(history["loss"]["val"][-1]))
    print("  точность на обучении: \t\t{:.2f} %".format(history["acc"]["train"][-1] * 100))
    print("  точность на валидации: \t\t{:.2f} %".format(history["acc"]["val"][-1] * 100))

    # Строим графики лосса и точности
    plot_histories([history], ["Model"])


def update_history(
    history: Dict[str, Dict[str, list]], loss: float, acc: float, num_batches: int, mode: str
) -> None:
    """
    Обновляет историю обучения, добавляя в нее loss и accuracy для текущей эпохи.

    Параметры:
        history (Dict[str, Dict[str, list]]): Словарь с историей обучения.
        loss (float): Суммарный loss за эпоху.
        acc (float): Суммарная точность за эпоху.
        num_batches (int): Количество батчей в эпохе.
        mode (str): Режим обучения ("train" или "val").
    """
    # Нормализуем loss и accuracy по количеству батчей
    loss /= num_batches
    acc /= num_batches
    history["loss"][mode].append(loss)
    history["acc"][mode].append(acc)


def get_batch_loss(
    X_batch: torch.Tensor,
    y_batch: torch.Tensor,
    model: nn.Module,
    criterion: Callable,
    current_loss: float,
    current_acc: float,
) -> Tuple[torch.Tensor, float, float]:
    """
    Вычисляет loss и accuracy для одного батча (без backward pass).

    Параметры:
        X_batch (torch.Tensor): Батч входных данных.
        y_batch (torch.Tensor): Батч меток.
        model (nn.Module): Модель для обучения (выдает логиты).
        criterion (Callable): Функция потерь.
        current_loss (float): Текущий лосс за батч.
        current_acc (float): Текущая точность за батч.

    Возвращает:
        Tuple[torch.Tensor, float, float]: Лосс на батче, обновленные суммарные loss и accuracy.
    """
    # Перемещаем данные на устройство (GPU/CPU)
    X_batch = X_batch.to(device)
    y_batch = y_batch.to(device)

    # Получаем логиты от модели
    logits = model(X_batch)

    # Вычисляем loss
    loss = criterion(logits, y_batch.long().to(device))

    # Сохраняем лоссы и точность
    current_loss += loss.detach().cpu().numpy()
    y_pred = logits.max(1)[1].detach().cpu().numpy()
    current_acc += np.mean(y_batch.cpu().numpy() == y_pred)
    return loss, current_loss, current_acc


def train(
    model: nn.Module,
    criterion: Callable,
    optimizer: optim.Optimizer,
    train_batch_gen: DataLoader,
    val_batch_gen: DataLoader,
    num_epochs: int = 40,
) -> Dict[str, Dict[str, list]]:
    """
    Обучает модель и возвращает историю обучения.

    Параметры:
        model (nn.Module): Модель для обучения.
        criterion (Callable): Функция потерь.
        optimizer (optim.Optimizer): Оптимизатор.
        train_batch_gen (DataLoader): Генератор батчей для обучения.
        val_batch_gen (DataLoader): Генератор батчей для валидации.
        num_epochs (int): Количество эпох (по умолчанию 40).

    Возвращает:
        Dict[str, Dict[str, list]]: История обучения, содержащая loss и accuracy для обучения и валидации.
    """
    history = defaultdict(lambda: defaultdict(list))

    for epoch in range(num_epochs):
        train_loss, val_loss = 0, 0
        train_acc, val_acc = 0, 0
        start_time = time.time()

        # ----------------------   ОБУЧЕНИЕ   ----------------------#
        model.train(True)

        # На каждой "эпохе" делаем полный проход по данным
        for X_batch, y_batch in train_batch_gen:

            # Вычисляем loss и accuracy для батча
            loss, train_loss, train_acc = get_batch_loss(
                X_batch, y_batch, model, criterion, train_loss, train_acc
            )

            # Обратный проход, вычисление градиентов
            loss.backward()
            # Шаг градиентного спуска для обновления оценок весов
            optimizer.step()
            # Зануление градиентов
            optimizer.zero_grad()

        # Обновляем историю обучения
        update_history(history, train_loss, train_acc, len(train_batch_gen), "train")

        # ----------------------   ВАЛИДАЦИЯ   ----------------------#
        model.train(False)

        # Контекстный менеджер, отключающий подсчет градиентов
        with torch.no_grad():
            # Полный проход по валидационному датасету
            for X_batch, y_batch in val_batch_gen:
                # Вычисляем loss и accuracy для батча
                _, val_loss, val_acc = get_batch_loss(
                    X_batch, y_batch, model, criterion, val_loss, val_acc
                )

        # Обновляем историю валидации
        update_history(history, val_loss, val_acc, len(val_batch_gen), "val")

        # Выводим информацию о текущей эпохе
        print_epoch(epoch, num_epochs, history, time.time() - start_time)
        history["time"]["epoch"].append(time.time() - start_time)

    return history
```

```python
# Лучшая модель, 99% accuracy и 597,946 параметров

model_A = nn.Sequential(
    nn.Conv2d(1, 16, kernel_size=7, padding=3), # (16, 28, 28)
    nn.BatchNorm2d(16),
    nn.ReLU(),

    nn.Conv2d(16, 16, kernel_size=9, padding=4),  # (16, 28, 28)
    nn.BatchNorm2d(16),
    nn.ReLU(),

    nn.Conv2d(16, 32, kernel_size=9, padding=4), #(32, 28, 28)
    nn.BatchNorm2d(32),
    nn.ReLU(),
    nn.MaxPool2d(2),# (32, 14, 14)

    nn.Conv2d(32, 32, kernel_size=7, padding=3),  # (32, 14, 14)
    nn.BatchNorm2d(32),
    nn.ReLU(),

    nn.Conv2d(32, 64, kernel_size=7, padding=3),  #  (64, 14, 14)
    nn.BatchNorm2d(64),
    nn.ReLU(),
    nn.MaxPool2d(2),    # - (64, 7, 7)

    nn.Conv2d(64, 64, kernel_size=3, padding=1), # (64, 7, 7)
    nn.BatchNorm2d(64),
    nn.ReLU(),

    nn.Flatten(),
    nn.Linear(64 * 7 * 7, 128),
    nn.ReLU(),
    nn.Linear(128, 10)
)

# Модель с чуть меньшим accuracy и 75,818 параметрами
model_B = nn.Sequential(
    nn.Conv2d(1, 16, kernel_size=3, padding=1),   # (16, 28, 28)
    nn.BatchNorm2d(16),
    nn.ReLU(),

    nn.Conv2d(16, 32, kernel_size=5, padding=2),  # (32, 28, 28)
    nn.BatchNorm2d(32),
    nn.ReLU(),
    nn.MaxPool2d(2), # -> (32, 14, 14)

    nn.Flatten(),
    nn.Linear(32 * 14 * 14, 10)
)

model_A.to(device)
model_B.to(device)
```

Зафиксируем гипотезу $H_0$: "accuracy модели A не превосходит accuracy модели B"

Для гипотезы нужно смотреть нижнюю границу интервала.
То есть $H_0$: $\text{Accuracy}_B - \text{Accuracy}_A >= 0$

Это односторонняя гипотеза. Для нее нужно строить односторонний доверительный интервал.

Модели нельзя рассматривать, как независимые, потому, что они тестируются на одних и тех же объектах.
По этому я считаю не отдельно 2 Accuracy,а их покомпонентную разность: `delta = correct_1 - correct_2`.
Если смотреть формально: Если модели независимые, то ошибки первой модели ничего не должны говорить об ошибках второй:
$$P(\text{model 1 correct | model 2 correct}) = P(\text{model 1 correct})$$

Но:
- если пример простой -> обе модели угадают.
- если пример сложный или разметка неверная -> могут ошибиться.

Для построения доверительного интервала нам недостаточно знать одну оценку accuracy. Нам нужно знать ее распределение. Поэтому ниже я применяю бутстреп что бы получить некоторое приблежение к распределению.

```python
def print_interval_and_delta_aсruracy(model_1, model_2, images, labels):
  model_1.eval()
  model_2.eval()
  with torch.no_grad():
    # Берем предсказание с максимальной вероятностью
    preds_1 = torch.argmax(model_1(images), dim=1)
    preds_2 = torch.argmax(model_2(images), dim=1)

    correct_1 = (preds_1 == labels).float().cpu().numpy()
    correct_2 = (preds_2 == labels).float().cpu().numpy()

    # -1, если model_1 ошиблась, а model_2 сказала правду
    # 1, если model_2 ошиблась, а model_1 сказала правду
    # Иначе - 0
    delta = correct_1 - correct_2

  # Делаем бутстреп
  n_boot = 10000
  bootstrap_means = []
  for _ in range(n_boot):
    resample = np.random.choice(delta, size=len(delta), replace=True)
    bootstrap_means.append(resample.mean())

  interval = np.percentile(bootstrap_means, [5, 100])
  print(f"Разность Accuracy: {delta.mean():.4f}")
  print(f"односторонняя нижняя доверительная граница 95%: ({interval[0]:.4f}, infinity)")
```

```python
from torch.utils.data import Subset, DataLoader

# test_batch типа torchvision.datasets.MNIST
indeces = np.random.choice(len(test_dataset), 500, replace=False)
subset = Subset(test_dataset, indeces)
loader = DataLoader(subset, batch_size=500, shuffle=False)
images, labels = next(iter(loader))
images, labels = images.to(device), labels.to(device)

print_interval_and_delta_aсruracy(model_A, model_B, images, labels)
```

Теперь выберите тестовую подвыборку стратифицированно. Удалось ли уменьшить длину интервала?

```python
from sklearn.model_selection import train_test_split

indices = np.arange(len(test_dataset))
labels = np.array([test_dataset[i][1] for i in indices])

_, stratified_indices = train_test_split(
    indices,
    test_size=500,
    stratify=labels,
    random_state=42
)

subset = Subset(test_dataset, stratified_indices)
loader = DataLoader(subset, batch_size=500, shuffle=False)

images, labels = next(iter(loader))
images, labels = images.to(device), labels.to(device)

print_interval_and_delta_aсruracy(model_A, model_B, images, labels)
```

Пусть $\Delta = \text{Accuracy}_B - \text{Accuracy}_A$

$Var(\Delta) = \mathbb{E}(Var(\Delta|C)) + Var(\mathbb{E}(\Delta | C))$ где C - распределение классов в подвыборке.

Стратификация убивает второе слагаемое(или значительно уменьшает его) -> дисперсия меньше -> нижняя граница сдвигается вправо
